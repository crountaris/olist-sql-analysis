```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

conn = sqlite3.connect("../olist.db")
```

```python
query = open("../queries/01_monthly_revenue.sql").read()
monthly = pd.read_sql(query, conn)
monthly = monthly[(monthly["month"] >= "2017-01") & (monthly["month"] <= "2018-08")]

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(monthly["month"], monthly["revenue"], marker="o")
ax.set_title("Monthly revenue (delivered orders)")
ax.set_xlabel("Month")
ax.set_ylabel("Revenue (BRL)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
query = open("../queries/05_delivery_vs_reviews.sql").read()
delivery = pd.read_sql(query, conn)

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(delivery["delivery_bucket"], delivery["avg_review"], color="steelblue")
ax.set_title("Average review score by delivery time")
ax.set_xlabel("Delivery time")
ax.set_ylabel("Average review (1-5)")
ax.set_ylim(0, 5)
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
query = open("../queries/03_top_categories.sql").read()
categories = pd.read_sql(query, conn)

fig, ax = plt.subplots(figsize=(9, 5))
ax.barh(categories["category"], categories["revenue"], color="seagreen")
ax.invert_yaxis()
ax.set_title("Top 10 product categories by revenue")
ax.set_xlabel("Revenue (BRL)")
plt.tight_layout()
plt.show()
```
